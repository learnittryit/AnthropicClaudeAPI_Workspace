```python
## Putting all together and call API
messages = []

add_user_messages(messages,"Give the 5 best dialogs of all times in movies")
print(messages)

stream_create_function = clientfunc.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True)

## Note we have used Stream function and not create function to get the stream response
with clientfunc.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:## Iterating the stream response
    for text in stream.text_stream:
        print(text,end="") ##### Important to note we are using text_stream and not event now
```
