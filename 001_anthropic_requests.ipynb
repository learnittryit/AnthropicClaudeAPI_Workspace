#### Earlier there was no context carried , but now it will be carried. [Keeping context in the conversation- histry of conversation is maintained and model can refer to it]
## Helper Functions 

##### function 1 - user message context 
#### function 2 - assistant message context. ( This is important to add , so the context via conv history is maaintained)
#### function 3 - final message returned

```python

## Helper Functions 

## function 1 - user message context 
## function 2 - assistant message context
## function 3 - final message returned

## funcion 1 
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

## funcion 2
def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

## funcion 3 
def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

## Using Helper Functions
##### Make a starting list of messages
##### Add initial question
##### Pass list of message to chat function to ger answer
####  Add assistance message to maintain the conversation history
##### Add followup ques
##### Call chat Again

```python
## Using Helper Functions

##### Make a starting list of messages
messages=[]
##### Add initial question
add_user_message(messages,"Which is a good book on emotional intelligence?")
print(messages)
##### Pass list of message to chat function to ger answer
answer = chat(messages)
print(answer)
#### Add assistant response to message list
add_assistant_message(messages, answer)
print(messages)
##### Add followup question
add_user_message(messages,"What is the main idea of the book?")
print(messages)
##### Call chat Again
chat(messages)
```
