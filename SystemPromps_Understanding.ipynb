## Math Tutor

```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
clientfunc = Anthropic()
model ="claude-sonnet-4-0"
```

```python
## user func
def add_user_messages(messages,text):
    user_message = {"role":"user","content":text}
    messages.append(user_message)

## assistant func
def add_assistant_messages(messages,text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)

## chat func
def chat(combinesMessages):
    message = clientfunc.messages.create(
        model=model,
        max_tokens=1000,
        messages=combinesMessages)
    return message.content[0].text

## chat func with system prompt
def chat_with_system_prompt(combinesMessages,system_prompt):

    message = clientfunc.messages.create(
        model=model,
        max_tokens=1000,
        messages=combinesMessages,
        system=system_prompt)
    return message.content[0].text
```

```python
## Putting all together and call API
messages = []

system_prompt = """
Do not directly answer the students question.
Guide them step by step.
"""

add_user_messages(messages,"How to solve 3x+5=20?")
print(messages)

response = chat(messages)
print(response)

print("\n-------------------------------------------\n")
response_with_system_prompt = chat_with_system_prompt(messages,system_prompt)
print(response_with_system_prompt)
```
