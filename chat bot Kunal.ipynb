## Make a sample chat bot

steps
1. load api key to hit API <br>
2. make 3 essential functions <br>
    func1 - user message <br>
    func2 - assistant message <br>
    func3 - chat and call model <br>
    

```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
clientfunc = Anthropic()
model ="claude-sonnet-4-0"
```

```python
## user func
def add_user_messages(messages,text):
    user_message = {"role":"user","content":text}
    messages.append(user_message)

## assistant func
def add_assistant_messages(messages,text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)

## chat func
def chat(combinesMessages):
    message = clientfunc.messages.create(
        model=model,
        max_tokens=1000,
        messages=combinesMessages)
    return message.content[0].text
```

```python
## Putting all together and call API
messages = []

add_user_messages(messages,"what is best quote on life? ")
print(messages)

response = chat(messages)
print(response)

## generate context by appending assistent message to messages list
add_assistant_messages(messages,response)
print(messages)

## user followup question
add_user_messages(messages,"what is best quote on self development?")
print(messages)

finalresponse = chat(messages)
print(finalresponse)
```
