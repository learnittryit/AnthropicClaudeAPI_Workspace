## Stream = Stream of events i.e basically an Iterator

steps
1. load api key to hit API <br>
2. make 3 essential functions <br>
    func1 - user message <br>
    func2 - assistant message <br>
    func3 - chat and call model <br>
3. <b> stream = chat function with parameter as stream , then iterate each stream 

    

```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
clientfunc = Anthropic()
model ="claude-sonnet-4-0"
```

```python
## user func
def add_user_messages(messages,text):
    user_message = {"role":"user","content":text}
    messages.append(user_message)

## assistant func
def add_assistant_messages(messages,text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)

## chat func
def chat(combinesMessages,temp=0.0):
    print("temperature is ",temp)
    message = clientfunc.messages.create(
        model=model,
        max_tokens=1000,
        messages=combinesMessages,
        temperature=temp)
    return message.content[0].text
```

```python
## Putting all together and call API
messages = []

add_user_messages(messages,"Give list of important online short ccertificates in demand offered by cloud providers, all cources must be less than 6 hours")
print(messages)

stream = clientfunc.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True)

## Iterating the stream response
for event in stream:
    print(event)

```

```python
### Testing response without streaming
response = chat(messages,temp=1.0)
print(response)
```
