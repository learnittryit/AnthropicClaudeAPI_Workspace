```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
clientfunc = Anthropic()
model ="claude-sonnet-4-0"
```

```python
## user func
def add_user_messages(messages,text):
    user_message = {"role":"user","content":text}
    messages.append(user_message)

## assistant func
def add_assistant_messages(messages,text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)

def chat(messages,system_prompt=None):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages        
    }

    if system_prompt:
        params["system"]=system_prompt
    
    message = clientfunc.messages.create(**params)        
    return message.content[0].text
```

```python
## Putting all together and call API
messages = []

system_prompt = """
Do not directly answer the students question.
Guide them step by step.
"""

add_user_messages(messages,"How to solve 3x+5=20?")
print(messages)

print("----------------------------------------\n WithOut system prompt \n----------------------------------------\n")
response = chat(messages)
print(response)

print("----------------------------------------\n With system prompt \n----------------------------------------\n")

response = chat(messages,system_prompt)
print(response)
```
